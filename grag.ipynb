```python
import graphrag.api as api
from graphrag.index.typing import PipelineRunResult
from graphrag.config.create_graphrag_config import create_graphrag_config
import yaml
import asyncio

async def main():
    # Load configuration from YAML
    settings = yaml.safe_load(open("grag/settings.yaml"))
    graphrag_config = create_graphrag_config(values=settings, root_dir="grag")

    # Build the index using the configured LLM (Ollama in this case)
    index_result: list[PipelineRunResult] = await api.build_index(config=graphrag_config)

    # Iterate through the workflow results and print their status
    for workflow_result in index_result:
        status = f"error\n{workflow_result.errors}" if workflow_result.errors else "success"
        print(f"Workflow Name: {workflow_result.workflow}\tStatus: {status}")

# Execute the main function
import sys

if sys.platform.startswith('win') and sys.version_info >= (3, 8):
    asyncio.set_event_loop_policy(asyncio.WindowsSelectorEventLoopPolicy())

asyncio.run(main())
```

```python
import graphrag.api as api
from graphrag.config.create_graphrag_config import create_graphrag_config
import yaml

settings = yaml.safe_load(open("grag/settings.yaml"))
graphrag_config = create_graphrag_config(values=settings, root_dir="grag")

index_result = await api.generate_indexing_prompts(
    config=graphrag_config,
    root="grag/",  # Adjust this path to the root directory of your documents
    chunk_size=512,
    limit=15,  # Limit must be a positive integer
    selection_method="random",  # DocSelectionType.RANDOM is represented as "random"
    domain=None,  # Set to None or specify a domain, e.g., "general"
    language="en",  # Specify the language or leave it as None
    max_tokens=1000,
    discover_entity_types=True,
    min_examples_required=3,
    n_subset_max=100,
    k=10,
)

entity_extraction = index_result[0]
summarization = index_result[1]
community = index_result[2]
```
